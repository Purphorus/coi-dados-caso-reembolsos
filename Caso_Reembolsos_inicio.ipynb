```python
from pathlib import Path
from urllib.request import urlretrieve
import pandas as pd

REPO = "https://raw.githubusercontent.com/Purphorus/coi-dados-caso-reembolsos/main/"
ARQUIVOS = ["briefing.md", "log_reembolsos.csv", "dicionario_codigos.xlsx", "quadro_pessoal.xlsx",
            "fluxograma_central.png"]
PASTA = Path(".")          # a pasta onde estão os arquivos do caso
for nome in ARQUIVOS:
    if not (PASTA / nome).exists():
        urlretrieve(REPO + nome, PASTA / nome)
        print("baixado:", nome)

print((PASTA / "briefing.md").read_text(encoding="utf-8"))
```

```python
log = pd.read_csv(PASTA / "log_reembolsos.csv", sep=";", dtype={"codigo": str})
dicionario = pd.read_excel(PASTA / "dicionario_codigos.xlsx", skiprows=2, dtype={"codigo": str})
equipe = pd.read_excel(PASTA / "quadro_pessoal.xlsx")
print(f"{len(log):,} linhas no log")
log.head(10)
```
